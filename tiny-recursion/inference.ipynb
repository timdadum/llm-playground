
# LLM Playground

When models are trained, this notebook can be used to empirically test, try and break the model. First, load the model using:

```python
from model import ModelConfig, BasicGPT, TRMModel
from transformers import GPT2TokenizerFast
import torch

CHECKPOINT_NAME = "trm-basic-100k-3e-4n2T-d512-K8"
CKPT_DIR = f"./checkpoints/{CHECKPOINT_NAME}"
MODEL_TYPE = "trm" # or "gpt"

device = "cuda" if torch.cuda.is_available() else "cpu"
config = ModelConfig()
model = TRMModel.from_pretrained(CKPT_DIR).to(device) if MODEL_TYPE == "trm" else BasicGPT.from_pretrained(CKPT_DIR).to(device)
model.eval()
tokenizer = GPT2TokenizerFast.from_pretrained("gpt2")
tokenizer.pad_token = tokenizer.eos_token  # pad = eos (id 50256)
```

```python
sequence = (
    "England is a country that is part of the United Kingdom. "
    "It shares land borders with Scotland to the north, "
)
inputs = tokenizer(sequence, return_tensors="pt").to(device)

with torch.no_grad():
    out_ids = model.generate(
        **inputs,
        max_new_tokens=128,
        do_sample=True,
        temperature=0.8,
        eos_token_id=tokenizer.eos_token_id,
        pad_token_id=tokenizer.eos_token_id,
    )
print(tokenizer.decode(out_ids[0], skip_special_tokens=True))
```
